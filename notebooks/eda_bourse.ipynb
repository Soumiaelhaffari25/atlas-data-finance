```python
from selenium import webdriverfrom selenium.webdriver.common.by import Byfrom selenium.webdriver.support.ui import WebDriverWait, Selectfrom selenium.webdriver.support import expected_conditions as ECimport pandas as pdfrom pathlib import Pathimport timeimport glob
```

```python
TICKERS = ["ATW", "IAM", "BCP"]download_dir = str((Path("../data/raw/bourse")).resolve())Path(download_dir).mkdir(parents=True, exist_ok=True)
```

```python
options = webdriver.ChromeOptions()prefs = {"download.default_directory": download_dir}options.add_experimental_option("prefs", prefs)driver = webdriver.Chrome(options=options)
```

```python
def download_instrument(ticker):    driver.get("https://www.casablanca-bourse.com/fr/instruments")    wait = WebDriverWait(driver, 20)    search_box = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "input[type='search'], input.form-control")))    search_box.clear()    search_box.send_keys(ticker)    time.sleep(2)    result = wait.until(EC.element_to_be_clickable((By.PARTIAL_LINK_TEXT, ticker)))    result.click()    time.sleep(2)    telecharger_btn = wait.until(EC.element_to_be_clickable((By.XPATH, "//*[contains(text(), 'lécharger')]")))    telecharger_btn.click()    time.sleep(5)
```

```python
for ticker in TICKERS:    download_instrument(ticker)driver.quit()
```

```python
files = sorted(glob.glob(f"{download_dir}/Cours_*.csv"))files
```

```python
dfs = []for f in files:    df = pd.read_csv(f, sep=";", encoding="utf-8-sig")    dfs.append(df)df_bourse = pd.concat(dfs, ignore_index=True)df_bourse.shape
```
